```python
#W8A16
class Shape:
    def area(self):
        pass
class rectangle(Shape):
    def __init__(self,w,h):
        self.w=w
        self.h=h
    def area(self):
        return self.w*self.h
class circle(Shape):
    def __init__(self,r):
        self.r=r
    def area(self):
        return 3.14*self.r*self.r
type=input().strip().lower()
if type=="rectangele":
    w,h=map(float,input().split())
    shape=rectangle(w,h)
elif type=="circle":
    r=int(input())
    shape=circle(r)
print(f"{shape.area():.2f}")
```

```python
#W8A18
class polygon:
    def poly(self):
        pass
    def peri(self):
        pass
class triangle(polygon):
    def __init__(self, a,b,c):
        self.a=a
        self.b=b
        self.c=c
    def peri(self):
        return self.a+self.b+self.c
a,b,c=map(float,input().split())
obj=triangle(a,b,c)
print(obj.peri())
```

```python
#W8A19
class person:
    def __init__ (self,name,age):
        self.name=name
        self.age=age
    def desi(self):
        return f"{self.name} {self.age}"
class student(person):
    def __init__(self,name,age,school):
        super().__init__(name,age)
        self.school=school
    def desi(self):
        return f"{self.name} {self.age} {self.school}"
type=input().strip().lower()
if type =="person":
    data=input().split()
    a1=data[0]
    a2=data[1]
    s1=person(a1,a2)
    print(s1.desi())
elif type=="student":
    data=input().split()
    a1=data[0]
    a2=data[1]
    a3=data[2]
    s1=student(a1,a2,a3)
    print(s1.desi())
```

```python
#W8A20
class account:
    def __init__(self,balance):
        self.balance=float(balance)
    def deposit(self,x):
        self.balance+=float(x)
    def getbalance(self):
        return self.balance
class savingaccount(account):
    def __init__(self,balance,rate):
        super().__init__(balance)
        self.rate=float(rate)
    def addinter(self):
        abc = self.balance * self.rate
        self.balance+=abc
if __name__=="__main__":
    data=input().split()
    acc =savingaccount(*data)
    acc.addinter()
    final=acc.getbalance()
    print( f"{final:.2f}")
        
```

```python
#W8A22
class matran:
    def __init__(self,a,b,c,d):
        self.a=a
        self.b=b
        self.c=c
        self.d=d
    def vet(self):
        return self.a + self.d
    def dinhthuc(self):
        return (self.a*self.d) - (self.b*self.c)
a,b,c,d=map(int,input().split())
res=matran(a,b,c,d)
tuan=res.vet()
tu=res.dinhthuc()
print(tuan,tu)
```

```python
#W8A24
class poly:
    def __init__(self,a,b,c):
        self.a=int(a)
        self.b=int(b)
        self.c=int(c)
    def evalue(self,x):
        val=int(x)
        return self.a*(val**2)+self.b*(val)+self.c
data=input().split()
obj = poly(data[0],data[1],data[2])
res=obj.evalue(data[3])
print(res)
```

```python
#W8A25
class poly:
    def __init__(self,a,b,c,d):
        self.a=int(a)
        self.b=int(b)
        self.c=int(c)
        self.d=int(d)
    def sum(self):
        if self.b+self.d>=0:
            return f"{self.a+self.c}x+{self.b+self.d}"
        else:
            return f"{self.a+self.c}-{abs(self.b+self.d)}"
data=input().split()
obj=poly(*data)
print(obj.sum())
    
```

```python
#W8A27
class product:
    def __init__(self,name,price):
        self.name=name
        self.price=float(price)
    def finalprice(self):
        pass
class discount(product):
    def __init__(self,name,price,percent):
        super().__init__(name,price)
        self.percent=float(percent)
    def finalprice(self):
        res=self.price*(100-self.percent)/100
        return f"{res:.2f}"
data=input().strip().split()
obj=discount(*data)
print(obj.finalprice())
```

```python
#W8A28
class counter:
    def __init__(self):
        self.count=0
    def inc(self):
        self.count+=1
    def dec(self):
        self.count-=1
    def value(self):
        return self.count
n=int(input().strip())
data=input().strip().split()
count=counter()
for i in data:
    if i =="dec":
        count.dec()
    elif i=="inc":
        count.inc()
print(count.value())
```

```python
#W8A29
class time:
    def __init__(self,h,m,s):
        self.h=int(h)
        self.m=int(m)
        self.s=int(s)
    def toseconds(self):
        return self.h*3600+self.m*60+self.s
data=input().strip().split()
obj=time(*data)
print(obj.toseconds())
```

```python
#W8A30
class stack:
    def __init__(self):
        self.item = []
    def push(self,x):
        self.item.append(x)
    def pop(self):
        self.item.pop()
    def top():
        return self.item[-1]
import sys
input=sys.stdin.read
data=input().splitlines()
if data:
    n=int(data[0].strip())
    mystack=stack()
    for i in range(1,n+1):
        if i >=len(data):
            break
        command=data[i].strip().split()
        if not command :
            continue
        action=command[0]
        if action=="push":
            value=command[1]
            mystack.push(value)
        elif action=='pop':
            mystack.pop()
        elif action=='top':
            print(mystack.top())

    
```

```python
#W7A17
n = int(input())
lst = list(map(int,input().split()))

dic = {}
for i in lst:
    if i not in dic:
        dic[i] = 1
    else:
        dic[i] += 1

target = n/2
for key in dic:
    if dic[key] >= target:
        print(key)
```

```python
#W9A16
def generate_binary(n, m):
    result = []
    def backtrack(current_str, count0, count1):
        if count0 == n and count1 == m:
            result.append(current_str)
            return
        if count0 < n:
            backtrack(current_str + '0', count0 + 1, count1)
        # Sau đó mới thử thêm '1'
        if count1 < m:
            backtrack(current_str + '1', count0, count1 + 1)
    backtrack("", 0, 0)
    return result
n, m = 2,2
for s in generate_binary(n, m):
    print(s)
```

```python
#W9A17
def longest_increasing_subsequence(arr):
    if not arr:
        return 0
    n = len(arr)
    dp = [1] * n  
    for i in range(1, n):
        for j in range(0, i):
            if arr[i] > arr[j]: 
                dp[i] = max(dp[i], dp[j] + 1)            
    return max(dp)
arr = list(map(int, input("Nhập dãy số (cách nhau bởi dấu cách): ").split()))
print("Độ dài dãy con tăng nghiêm ngặt dài nhất:", longest_increasing_subsequence(arr))
```

```python
#W9A18
def count_pairs_with_sum(arr, k):
    count = 0
    freq = {}  
    for x in arr:
        target = k - x
        if target in freq:
            count += freq[target]
            
        freq[x] = freq.get(x, 0) + 1
        
    return count

arr = list(map(int, input("Nhập dãy số: ").split()))
k = int(input("Nhập số k: "))
print("Số cặp thỏa mãn:", count_pairs_with_sum(arr, k))
```

```python
#W9A19
def remove_duplicates(arr):
    seen = set()
    result = []
    for item in arr:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result

arr = input("Nhập danh sách các phần tử: ").split()
print("Danh sách sau khi lọc trùng:", " ".join(remove_duplicates(arr)))
```

```python
#W9A20
def normalize_string(s):
    result = []
    in_word = False  
    for char in s:
        if char != ' ' and char != '\t': 
            if not in_word and len(result) > 0:
                result.append(' ')  
            result.append(char)
            in_word = True
        else:
            in_word = False
        
    is_first_letter = True
    for i in range(len(result)):
        char_code = ord(result[i])
        
        if result[i] == ' ':
            is_first_letter = True
        else:
            if is_first_letter:
                
                if 97 <= char_code <= 122:
                    result[i] = chr(char_code - 32)
                is_first_letter = False
            else:
                
                if 65 <= char_code <= 90:
                    result[i] = chr(char_code + 32)
                    
    return "".join(result)

# Nhập dữ liệu chạy thử
s = input("Nhập chuỗi cần chuẩn hóa: ")
print("Chuỗi sau khi chuẩn hóa:", normalize_string(s))
```

```python
#W9A21
def quickselect(arr, left, right, k_smallest):
    if left == right:
        return arr[left]
    pivot = arr[right]
    i = left
    
    for j in range(left, right):
        if arr[j] <= pivot:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1
            
    arr[i], arr[right] = arr[right], arr[i]
    
    if i == k_smallest:
        return arr[i]
    elif i > k_smallest:
        return quickselect(arr, left, i - 1, k_smallest)
    else:
        return quickselect(arr, i + 1, right, k_smallest)

def find_kth_largest(arr, k):
    n = len(arr)
    return quickselect(arr, 0, n - 1, n - k)
arr = list(map(int, input("Nhập dãy số: ").split()))
k = int(input("Nhập k (tìm số lớn thứ k): "))
print(f"Phần tử lớn thứ {k} là:", find_kth_largest(arr, k))
```

```python
#W9A22
def is_anagram(str1, str2):
    if len(str1) != len(str2):
        return False
        
    count = {}
    
    for char in str1:
        count[char] = count.get(char, 0) + 1
        
    for char in str2:
        if char not in count:
            return False
        count[char] -= 1
        if count[char] < 0:
            return False
            
    return True

s1 = input("Nhập chuỗi thứ nhất: ")
s2 = input("Nhập chuỗi thứ hai: ")

if is_anagram(s1, s2):
    print("Hai chuỗi LÀ anagram của nhau.")
else:
    print("Hai chuỗi KHÔNG PHẢI là anagram của nhau.")
```

```python
#W9A23
def max_number_k_digits(num_str, k):
    stack = []
    to_remove = len(num_str) - k
    
    for digit in num_str:
        while stack and stack[-1] < digit and to_remove > 0:
            stack.pop()
            to_remove -= 1
        stack.append(digit)
    
    result = stack[:k]
    
    return "".join(result)

num_str = "78934340999"
k = 7
print("Kết quả lớn nhất tìm được:", max_number_k_digits(num_str, k))
# Output: 9440999
```

```python
#W9A24
def compress_string(s):
    if not s:
        return ""
        
    result = []
    count = 1
    
    for i in range(len(s) - 1):
        if s[i] == s[i+1]:
            count += 1
        else:
            result.append(f"{s[i]}{count}")
            count = 1  
            
    
    return "".join(result)

s = "aaaaaafdhhhhh"
print("Chuỗi sau khi nén:", compress_string(s))
```

```python
#W9A25
def check_strong_password(password):
    if len(password) < 8:
        return False
        
    has_lower = False
    has_upper = False
    has_digit = False
    has_special = False
    
    special_characters = {'!', '@', '#', '$'}
    
    for char in password:
        if char.islower():
            has_lower = True
        elif char.isupper():
            has_upper = True
        elif char.isdigit():
            has_digit = True
        elif char in special_characters:
            has_special = True      
    return has_lower and has_upper and has_digit and has_special
pwd = input("Nhập mật khẩu cần kiểm tra: ")
if check_strong_password(pwd):
    print("Mật khẩu MẠNH!")
else:
    print("Mật khẩu YẾU! (Vui lòng kiểm tra lại độ dài, chữ hoa, chữ thường, chữ số và ký tự đặc biệt !, @, #, $)")
```
